```python
def hamming(data):
    n = len(data)
    r = 0
    # Находим количество контрольных битов
    while 2 ** r < n + r + 1:
        r += 1

    result = ['0'] * (n + r)
    j = 0

    # Расставляем информационные биты
    for i in range(1, n + r + 1):
        if i & (i - 1) != 0:
            result[i - 1] = data[j]
            j += 1

    # Вычисляем контрольные биты
    for p in range(r):
        pos = 2 ** p
        s = 0

        for i in range(1, n + r + 1):
            if i & pos:
                s ^= int(result[i - 1])

        result[pos - 1] = str(s)

    return ''.join(result)

# Ввод сообщения
message = input("Введите сообщение: ")

# Перевод символов в двоичный код
binary = ''.join(format(b, '08b') for b in message.encode('utf-8'))

print("Двоичное сообщение:", binary)

# Деление на блоки по 16 бит
blocks = [binary[i:i + 16] for i in range(0, len(binary), 16)]

encoded = []

for i, block in enumerate(blocks, 1):
    code = hamming(block)
    encoded.append(code)

    print(f"\nБлок {i}: {block}")
    print(f"Код Хэмминга: {code}")

# Итог
print("\nЗакодированное сообщение:")
print(''.join(encoded))

```
