```python
import math
import heapq


# 1. Минимальное количество бит
def zadanie1():
    n = int(input("Введите размер алфавита: "))

    if n <= 0:
        print("Размер алфавита должен быть положительным")
        return

    bits = math.ceil(math.log2(n))
    print("Минимальное количество бит:", bits)


# 2. Количество различных символов
def zadanie2():
    k = int(input("Введите длину кода в битах: "))

    if k < 0:
        print("Длина не может быть отрицательной")
        return

    result = 2 ** k
    print("Количество символов:", result)


# 3. Количество бит для сообщения
def zadanie3():
    n = int(input("Введите длину сообщения: "))
    m = int(input("Введите количество символов алфавита: "))

    if n < 0 or m <= 0:
        print("Некорректные данные")
        return

    freq = []

    for i in range(m):
        p = float(input(f"Частота символа {i + 1} (%): "))
        freq.append(p)

    if any(p < 0 for p in freq) or not math.isclose(sum(freq), 100):
        print("Сумма частот должна быть 100%")
        return

    # Равномерное кодирование
    bits = math.ceil(math.log2(m))
    total = n * bits

    print("Бит на один символ:", bits)
    print("Равномерный код:", total, "бит")

    # Кодирование Хаффмана
    heap = [p for p in freq if p > 0]
    heapq.heapify(heap)

    summa = 0

    if len(heap) > 1:
        while len(heap) > 1:
            a = heapq.heappop(heap)
            b = heapq.heappop(heap)

            s = a + b
            summa += s
            heapq.heappush(heap, s)

        average = summa / 100
    else:
        average = 1

    print("Средняя длина кода Хаффмана:", round(average, 4))
    print("По Хаффману примерно:", round(n * average), "бит")


# 4. Сравнение двух кодов
def zadanie4():
    n = int(input("Введите количество символов: "))

    freq = []
    code1 = []
    code2 = []

    for i in range(n):
        p = float(input(f"Частота символа {i + 1} (%): "))
        freq.append(p)

        a = input("Код в варианте A: ")
        b = input("Код в варианте B: ")

        code1.append(a)
        code2.append(b)

    if not math.isclose(sum(freq), 100):
        print("Сумма частот должна быть 100%")
        return

    avg1 = 0
    avg2 = 0

    for i in range(n):
        avg1 += freq[i] / 100 * len(code1[i])
        avg2 += freq[i] / 100 * len(code2[i])

    print("Средняя длина кода A:", round(avg1, 4))
    print("Средняя длина кода B:", round(avg2, 4))

    if avg1 < avg2:
        print("Код A эффективнее")
    elif avg2 < avg1:
        print("Код B эффективнее")
    else:
        print("Коды одинаково эффективны")


# 5. Неравенство Крафта
def zadanie5():
    lengths = list(map(int, input(
        "Введите длины кодовых слов через пробел: "
    ).split()))

    if not lengths or any(x <= 0 for x in lengths):
        print("Длины должны быть положительными")
        return

    kraft = sum(2 ** (-x) for x in lengths)

    print("Количество заданных кодовых слов:", len(lengths))
    print("Сумма Крафта:", round(kraft, 6))

    if kraft <= 1:
        print("Такой префиксный код возможен")

        max_len = max(lengths)
        used = sum(2 ** (max_len - x) for x in lengths)
        free = 2 ** max_len - used

        print("Дополнительных слов длины", max_len, ":", free)
        print("Всего можно иметь:", len(lengths) + free)
    else:
        print("Префиксный код невозможен")


# Главное меню
while True:
    print("\n===== ВАРИАНТ 9 =====")
    print("1 - Минимальное количество бит")
    print("2 - Количество символов")
    print("3 - Объем сообщения")
    print("4 - Сравнение кодов")
    print("5 - Неравенство Крафта")
    print("0 - Выход")

    choice = input("Выберите задание: ")

    if choice == "1":
        zadanie1()
    elif choice == "2":
        zadanie2()
    elif choice == "3":
        zadanie3()
    elif choice == "4":
        zadanie4()
    elif choice == "5":
        zadanie5()
    elif choice == "0":
        print("Программа завершена")
        break
    else:
        print("Неверный выбор")
789
```
